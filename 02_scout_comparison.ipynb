```python
s1_r1 = pyXLMS.transform.unique(s1_r1, by="protein")
```

# **Figures**

```python
import os

if not os.path.exists("plots"):
    os.makedirs("plots")

if not os.path.exists("plots/scout_comparison"):
    os.makedirs("plots/scout_comparison")
```

```python
import numpy as np
```

```python
scout_1_csms = [
    len(s1_r1["crosslink-spectrum-matches"]),
    len(s1_r2["crosslink-spectrum-matches"]),
    len(s1_r3["crosslink-spectrum-matches"]),
]
print(", ".join([str(x) for x in scout_1_csms]))
```

```python
scout_2_csms = [
    len(s2_r1["crosslink-spectrum-matches"]),
    len(s2_r2["crosslink-spectrum-matches"]),
    len(s2_r3["crosslink-spectrum-matches"]),
]
print(", ".join([str(x) for x in scout_2_csms]))
```

```python
scout_1_xls = [
    len(s1_r1["crosslinks"]),
    len(s1_r2["crosslinks"]),
    len(s1_r3["crosslinks"]),
]
print(", ".join([str(x) for x in scout_1_xls]))
```

```python
scout_2_xls = [
    len(s2_r1["crosslinks"]),
    len(s2_r2["crosslinks"]),
    len(s2_r3["crosslinks"]),
]
print(", ".join([str(x) for x in scout_2_xls]))
```

```python
import matplotlib.pyplot as plt

reps = ("R1", "R2", "R3")
se = {
    "Scout 2.0.0 beta Liu": scout_1_csms,
    "Scout 2.1.0 Borges": scout_2_csms,
}

x = np.arange(len(reps))
width = 0.4
multiplier = 0

fig, ax = plt.subplots(layout="constrained", figsize=(7, 4))

colors = ["#6d4bff", "#ac99ff"]
for attribute, measurement in se.items():
    offset = width * multiplier
    rects = ax.bar(
        x + offset, measurement, width, label=attribute, color=colors[multiplier % 2]
    )
    ax.bar_label(rects, padding=14, fmt="{:.0f}")
    multiplier += 1

ax.set_xlabel("Replicate")
ax.set_ylabel("Number of CSMs at estimated 1% FDR")
ax.set_title("Number of Crosslink-Spectrum-Matches (CSMs) by Scout version")
ax.set_xticks(x + width / 2, reps)
ax.legend(loc="upper right", ncols=1)
ax.set_ylim(0, 500)
plt.savefig(
    "plots/scout_comparison/csms.svg",
    dpi=300,
    transparent=True,
    bbox_inches="tight",
)
```

```python
reps = ("R1", "R2", "R3")
se = {
    "Scout 2.0.0 beta Liu": scout_1_xls,
    "Scout 2.1.0 Borges": scout_2_xls,
}

x = np.arange(len(reps))
width = 0.4
multiplier = 0

fig, ax = plt.subplots(layout="constrained", figsize=(7, 4))

colors = ["#6d4bff", "#ac99ff"]
for attribute, measurement in se.items():
    offset = width * multiplier
    rects = ax.bar(
        x + offset, measurement, width, label=attribute, color=colors[multiplier % 2]
    )
    ax.bar_label(rects, padding=14, fmt="{:.0f}")
    multiplier += 1

ax.set_xlabel("Replicate")
ax.set_ylabel("Number of XLs at estimated 1% FDR")
ax.set_title("Number of Crosslinks (XLs) by Scout version")
ax.set_xticks(x + width / 2, reps)
ax.legend(loc="upper right", ncols=1)
ax.set_ylim(0, 500)
plt.savefig(
    "plots/scout_comparison/xls.svg",
    dpi=300,
    transparent=True,
    bbox_inches="tight",
)
```
